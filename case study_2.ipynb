```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5",
                "S6", "S7", "S8", "S9", "S10"],

    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],

    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],

    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],

    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],

    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}
df = pd.DataFrame(data)
print("Student Performance Dataset:")
print(df)
print("\nNumber of students:")
print(len(df))
print("\nColumn names:")
print(df.columns)
print("\nData types:")
print(df.dtypes)
print("\nFirst five records:")
print(df.head())

print("\nLast five records:")
print(df.tail())

```

```python
df = pd.DataFrame(data)
print("Complete Dataset:")
print(df)
print("\nAverage Study Hours:", df["Study_Hours"].mean())
print("Average Attendance:", df["Attendance"].mean())
print("Average Assignment Score:", df["Assignment_Score"].mean())
print("Average Previous Score:", df["Previous_Score"].mean())
print("Average Final Score:", df["Final_Score"].mean())
highest = df.loc[df["Final_Score"].idxmax()]
print("\nStudent with the highest Final Score:")
print(highest)
lowest = df.loc[df["Final_Score"].idxmin()]
print("\nStudent with the lowest Final Score:")
print(lowest)
print("\nStudents with Attendance greater than 80%:")
print(df[df["Attendance"] > 80])
print("\nStudents with Final Score greater than 80:")
print(df[df["Final_Score"] > 80])
print("\nStatistical Summary:")
print(df.describe())
```

```python
final_scores = df["Final_Score"].to_numpy()
mean_score = np.mean(final_scores)
median_score = np.median(final_scores)
maximum_score = np.max(final_scores)
minimum_score = np.min(final_scores)
standard_deviation = np.std(final_scores)
print("Final Scores:", final_scores)
print("Mean:", mean_score)
print("Median:", median_score)
print("Maximum:", maximum_score)
print("Minimum:", minimum_score)
print("Standard Deviation:", standard_deviation)
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(df["Study_Hours"], df["Final_Score"], color="blue")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Study Hours vs Final Score")
plt.grid(True)
plt.show()

plt.figure(figsize=(7, 5))
plt.scatter(df["Attendance"], df["Final_Score"], color="green")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Score")
plt.title("Attendance vs Final Score")
plt.grid(True)
plt.show()

plt.figure(figsize=(7, 5))
plt.hist(df["Final_Score"], bins=5, color="orange", edgecolor="black")
plt.xlabel("Final Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Final Scores")
plt.grid(axis="y")
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Final_Score"], color="purple")
plt.xlabel("Student")
plt.ylabel("Final Score")
plt.title("Student-wise Final Scores")
plt.grid(axis="y")
plt.show()

highest_score = df["Final_Score"].max()
lowest_score = df["Final_Score"].min()

highest_student = df.loc[df["Final_Score"].idxmax(), "Student"]
lowest_student = df.loc[df["Final_Score"].idxmin(), "Student"]

print("Highest Score:", highest_score, "by", highest_student)
print("Lowest Score:", lowest_score, "by", lowest_student)
```
