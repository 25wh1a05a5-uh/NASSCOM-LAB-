**Exception handling and debugging**

```python
try:
    x = int(input("Enter a number: "))
    print("You entered:", x)
except ValueError:
    print("That was not a valid number!")
```

```python
try:
    num = int(input("Enter a number to divide 100: "))
    result = 100 / num    # pdb.set_trace()  # Start interactive debugger here
except ZeroDivisionError:
    print("Cannot divide by zero!")
except ValueError:
    print("Invalid input! Not a number.")
else:
    print("Result is:", result)
finally:
    print("Execution complete. This block always runs.")
```

```python
try:
    a = int("ten")
    b = 10 / 0
except (ValueError, ZeroDivisionError) as e:
    print("An error occurred:", e)
```

```python
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    else:
        print("Age is valid.")

try:
    check_age(-5)
except ValueError as e:
    print("Caught an exception:", e)
```

```python
import os
import datetimeprint("Total Temperature Readings:", len(temperature_readings))
import random

DEVICE_TYPES = [
    "Temperature_Sensor",
    "Humidity_Sensor",
    "Pressure_Sensor",
    "Light_Sensor"
]

LOCATIONS = [
    "Lab1",
    "Lab2",
    "OfficeA",
    "OfficeB",
    "Warehouse"
]

READING_RANGES = {
    "Temperature_Sensor": (18.0, 30.0),
    "Humidity_Sensor": (30.0, 80.0),
    "Pressure_Sensor": (980.0, 1030.0),
    "Light_Sensor": (100, 1000)
}

print("\n--- Task 1: Data Generation and Basic Structures ---")
#task 1.1
raw_sensor_data = []

start_time = datetime.datetime.now()

for i in range(100):
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    minimum, maximum = READING_RANGES[device_type]
    value = round(random.uniform(minimum, maximum), 2)

    timestamp = (
        start_time + datetime.timedelta(minutes=i)
    ).strftime("%Y-%m-%d %H:%M:%S")

    device_id = f"DEV{i + 1:03d}"

    raw_sensor_data.append(
        (
            timestamp,
            device_id,
            device_type,
            location,
            value
        )
    )
print("\nTask 1.1 Output")
print("Total Sensor Readings:", len(raw_sensor_data))

for reading in raw_sensor_data:
    print(reading)

```

```python
temperature_readings = []

for reading in raw_sensor_data:
    temperature_readings.append({
        "timestamp": reading[0],
        "device_id": reading[1],
        "location": reading[3],
        "temperature_celsius": reading[4]
    })



for reading in temperature_readings:
    print(reading)
```

```python
unique_locations = set()

for reading in raw_sensor_data:
    unique_locations.add(reading[3])

print("Unique Locations:")

for location in unique_locations:
    print(location)

print("\nTotal Unique Locations:", len(unique_locations))
```

```python
device_stats = {}

for device in DEVICE_TYPES:
    device_stats[device] = (float("inf"), float("-inf"))

for reading in raw_sensor_data:
    device_type = reading[2]
    value = reading[4]

    minimum, maximum = device_stats[device_type]

    if value < minimum:
        minimum = value

    if value > maximum:
        maximum = value

    device_stats[device_type] = (minimum, maximum)



for device, stats in device_stats.items():
    print(f"{device}: Min = {stats[0]}, Max = {stats[1]}")
```

```python
def calculate_average_temperature(temp_data_list):
    if len(temp_data_list) == 0:
        return 0.0

    total = 0

    for reading in temp_data_list:
        total += reading["temperature_celsius"]

    average = total / len(temp_data_list)

    return round(average, 2)
average_temperature = calculate_average_temperature(temperature_readings)

print("\nTask 2.2 Output")
print("Average Temperature:", average_temperature)
```
